```python
#Imports
import pandas as pd
import numpy as np
import holoviews as hv
import os
from holoviews import opts

hv.extension('bokeh', 'matplotlib')
```

```python
# Task 1: Load site CSVs into dictionaries keyed by USGS site ID

# Self-contained imports (so this cell works even if run before the imports cell)
import os
import pandas as pd

# If notebook is run from getting_started/, this points to the local data folder
# (ID, UT, WY subfolders)
data_folder = "data"

ID_files = [f for f in os.listdir(os.path.join(data_folder, "ID")) if f.endswith(".csv")]
UT_files = [f for f in os.listdir(os.path.join(data_folder, "UT")) if f.endswith(".csv")]
WY_files = [f for f in os.listdir(os.path.join(data_folder, "WY")) if f.endswith(".csv")]

ID_dict = {f.split("_")[0]: pd.read_csv(os.path.join(data_folder, "ID", f)) for f in ID_files}
UT_dict = {f.split("_")[0]: pd.read_csv(os.path.join(data_folder, "UT", f)) for f in UT_files}
WY_dict = {f.split("_")[0]: pd.read_csv(os.path.join(data_folder, "WY", f)) for f in WY_files}

print(f"Loaded ID sites: {len(ID_dict)}")
print(f"Loaded UT sites: {len(UT_dict)}")
print(f"Loaded WY sites: {len(WY_dict)}")

print("\nExample keys:")
print("ID:", sorted(ID_dict.keys())[:2])
print("UT:", sorted(UT_dict.keys())[:6])
print("WY:", sorted(WY_dict.keys())[:6])
```

## Task 2 Efficiently clean the data

Much of the USGS data is not needed, or bunches of data may be missing. 

Make a function called ``clean_df()`` that takes in a dataframe, null columns, and a specific state and conducts the following actions:

* Removes the unwanted columns (e.g., create a variable of a list of undesireable columns such as 'variable', 'measurement_unit', 'qualifiers', 'series')
* Adds a column for state and fills the column with the states name
* Sets the date column to a pandaas datetime object (e.g., pd.to_datetime())


Apply the cleaning function to each dictionary of dataframes

```python
# Task 2: Clean each site's dataframe and tag by state

def clean_df(df, null_cols, state):
    cleaned = df.copy()
    drop_cols = [col for col in null_cols if col in cleaned.columns]
    cleaned = cleaned.drop(columns=drop_cols)
    cleaned["state"] = state
    if "Datetime" in cleaned.columns:
        cleaned["Datetime"] = pd.to_datetime(cleaned["Datetime"], errors="coerce")
    elif "datetime" in cleaned.columns:
        cleaned["datetime"] = pd.to_datetime(cleaned["datetime"], errors="coerce")
    elif "date" in cleaned.columns:
        cleaned["date"] = pd.to_datetime(cleaned["date"], errors="coerce")
    return cleaned
unwanted_cols = ["variable", "measurement_unit", "qualifiers", "series"]
ID_dict = {site_id: clean_df(df, unwanted_cols, "ID") for site_id, df in ID_dict.items()}
UT_dict = {site_id: clean_df(df, unwanted_cols, "UT") for site_id, df in UT_dict.items()}
WY_dict = {site_id: clean_df(df, unwanted_cols, "WY") for site_id, df in WY_dict.items()}
print("Task 2 complete.")
print("Example cleaned columns:", ID_dict[next(iter(ID_dict))].columns.tolist())
```

## Task 3: Assign a Location variable to each site

Google the USGS ID for each of the streamflow monitoring station IDs and create a new column called **Site_type** that fills each dataframe with how you idenitify the station: Headwater, Tailwater, Near_Terminus.

Once complete, combine all of the dictionary dataframes into one large DataFrame called **All_Streams**

Double check that the All_Streams DataFrame has all of your sites by running ``All_Streams['USGS_ID'].unique()``. All of your sites should be represented.

```python
# Task 3: Assign Site_type by USGS ID and combine into All_Streams
# Site_type identification below is based on official USGS station names and drainag
# ontext

# Headwater: smaller upstream mountain/upper-basin stations
# Tailwater: downstream of major confluence/dam/urban reaches
# Near_Terminus: lower mainstem stations near basin outlet

site_type_map = {
    # Idaho
    "13168500": "Tailwater",      # Bruneau R nr Hot Spring (lower mainstem reach)
    "13185000": "Headwater",      # Boise R nr Twin Springs (upper basin)
    "13206000": "Tailwater",      # Boise R at Glenwood Bridge nr Boise (downstream urban reach)
    "13249500": "Tailwater",      # Payette R nr Emmett (lower basin)
    "13251000": "Near_Terminus",  # Payette R nr Payette (near Snake confluence reach)
    "13317000": "Near_Terminus",  # Salmon R at White Bird (large lower mainstem)

    # Utah
    "09261000": "Near_Terminus",  # Green R nr Jensen (very large lower mainstem reach)
    "09266500": "Headwater",      # Ashley Ck nr Vernal (small tributary basin)
    "10136500": "Tailwater",      # Weber R at Gateway (mid/lower mainstem)
    "10163000": "Near_Terminus",  # Provo R at Provo (lower reach near Utah Lake)
    "10171000": "Tailwater",      # Jordan R at Salt Lake City (downstream urban corridor)
    "10215900": "Headwater",      # Manti Ck blw Dugway Ck (small mountain watershed)

    # Wyoming / Colorado group
    "06751490": "Headwater",      # N Fork Cache La Poudre at Livermore (upper basin)
    "06752280": "Near_Terminus",  # Cache La Poudre near Timnath (lower plains reach)
    "09234500": "Tailwater",      # Green R near Greendale (regulated/dam-influenced reach)
    "09255000": "Headwater",      # Slater Fork near Slater (small tributary basin)
    "10011500": "Headwater",      # Bear R near UT-WY state line (upper reach)
    "10039500": "Near_Terminus",  # Bear R at Border (large lower-basin reach of selected sites)
}

# Add site type 
def add_site_type_from_key(stream_dict, site_map):
    out = {}
    for site_id, df in stream_dict.items():
        updated = df.copy()
        if "USGS_ID" in updated.columns:
            updated["USGS_ID"] = updated["USGS_ID"].astype(str).str.zfill(8)
        updated["Site_type"] = site_map.get(site_id, "Unknown")
        out[site_id] = updated
    return out

ID_dict = add_site_type_from_key(ID_dict, site_type_map)
UT_dict = add_site_type_from_key(UT_dict, site_type_map)
WY_dict = add_site_type_from_key(WY_dict, site_type_map)

All_Streams = pd.concat(
    [*ID_dict.values(), *UT_dict.values(), *WY_dict.values()],
    ignore_index=True
    )

#Print the output
print("All_Streams shape:", All_Streams.shape)
print("Unique USGS IDs in All_Streams:")
print(sorted(All_Streams["USGS_ID"].astype(str).unique()))
print("\nSite type counts:")
print(All_Streams[["USGS_ID", "Site_type"]].drop_duplicates()["Site_type"].value_counts())
```

## Task 4 Make a Holoeviews object with dropdowns for USGS ID and State
The following code block should work and allow interactivity. Note, the USGS_ID and State should be dropdowns. If they are not, you will need to fix something...

For the plots to work, you will need to have the correct state selected for the respetive USGS_ID


```python
# Task 4 prep: ensure columns needed for interactive dropdowns exist
task4_df = All_Streams.copy()

# Keep station IDs as 8-char strings
task4_df["USGS_ID"] = task4_df["USGS_ID"].astype(str).str.zfill(8)

# Ensure datetime and flow are usable for plotting
task4_df["Datetime"] = pd.to_datetime(task4_df["Datetime"], errors="coerce")
task4_df["USGS_flow"] = pd.to_numeric(task4_df["USGS_flow"], errors="coerce")

# Build a title-case State column required by Task 4 prompt
task4_df["State"] = task4_df["state"].astype(str).str.upper()

# Remove rows that cannot be plotted
task4_df = task4_df.dropna(subset=["Datetime", "USGS_flow", "USGS_ID", "State"])

print("Task 4 input shape:", task4_df.shape)
print("States:", sorted(task4_df["State"].unique()))
```

```python
# Task 4: Make a HoloViews object with dropdowns for USGS ID and State
vdims = ["USGS_flow"]
kdims = ["USGS_ID", "Datetime", "State"]

def create_holoviews_object(df):
    ds = hv.Dataset(df, kdims=kdims, vdims=vdims)

    # Remaining key dimensions (USGS_ID and State) become interactive selectors
    layout = ds.to(hv.Curve, "Datetime", "USGS_flow").opts(
        opts.Curve(width=750, height=320, framewise=True, tools=["hover"])
    )
    return layout

task4_plot = create_holoviews_object(task4_df)
task4_plot
```
